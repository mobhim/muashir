## 1. قراءة الملف واستكشافه

الجدول في pandas اسمه `DataFrame`، والعادة نسميه `df`.

```python
import pandas as pd

fin = pd.read_csv("../data/annual_financials.csv")
companies = pd.read_csv("../data/companies.csv")

fin.head()   # أول 5 صفوف
```

```python
fin.info()   # الأعمدة وأنواعها وكم قيمة غير فارغة في كل عمود
```

لاحظ في ناتج `info()` أن بعض الأعمدة فيها أقل من 30 قيمة. هذي **قيم مفقودة** (`NaN`)، وهي شي طبيعي في البيانات الحقيقية. بنرجع لها في التمارين.

## 2. الفلترة

تكتب الشرط داخل أقواس مربعة. الشرط نفسه يرجّع عمود `True/False`، و pandas يبقي الصفوف اللي قيمتها `True`.

```python
# كل صفوف أرامكو
fin[fin["ticker"] == 2222]
```

```python
# شرطين مع بعض: استخدم & وحط كل شرط بين قوسين
fin[(fin["year"] == 2025) & (fin["revenue_m_sar"] > 50000)]
```

## 3. إضافة عمود محسوب

العملية تنطبق على العمود كله مرة وحدة، بدون `for`.

```python
# نسبة التوزيع من الربح = التوزيع للسهم ÷ ربحية السهم
fin["payout_ratio"] = fin["dividend_per_share"] / fin["eps"]
fin[["ticker", "year", "eps", "dividend_per_share", "payout_ratio"]].head(10)
```

## 4. التجميع `groupby`

تقسم الصفوف إلى مجموعات، وتحسب رقم واحد لكل مجموعة.

```python
# متوسط الإيرادات لكل شركة على الخمس سنوات
fin.groupby("ticker")["revenue_m_sar"].mean()
```

## 5. دمج جدولين `merge`

جدول النتائج فيه الرمز فقط، وجدول الشركات فيه الاسم والقطاع. ندمجهم على العمود المشترك `ticker`.

```python
df = fin.merge(companies, on="ticker")
df[["name_ar", "sector", "year", "revenue_m_sar"]].head()
```
